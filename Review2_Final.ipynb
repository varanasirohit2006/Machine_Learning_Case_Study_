### B1.3 — Missing Value Imputation (Median for Numerical, Mode for Categorical)

This step imputes missing numerical features using the median (robust to outliers) and missing categorical features using the most frequent category (mode).


````python
# Numerical columns → median
for column in numerical_features:
    if df_clean[column].isnull().sum() > 0:
        df_clean[column] = df_clean[column].fillna(
            df_clean[column].median()
        )

# Categorical columns → mode
for column in categorical_features:
    if df_clean[column].isnull().sum() > 0:
        df_clean[column] = df_clean[column].fillna(
            df_clean[column].mode()[0]
        )

print("Missing-value imputation completed.")
````

## B1.8 — Statistical Outlier Detection (IQR Method)

Potential outliers are evaluated using the 1.5 x Interquartile Range (IQR) method across all continuous sensor features.


````python
# IQR-based outlier count for every numerical feature

outlier_summary = []

for column in numerical_features:
    
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outlier_count = (
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ).sum()
    
    outlier_percentage = (
        outlier_count / len(df_clean) * 100
    )
    
    outlier_summary.append({
        "Feature": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": round(outlier_percentage, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    "Outlier Count",
    ascending=False
)
````

### B1.11 — Outlier Retention & Treatment Justification

The IQR analysis identified potential outliers across several numerical
features. The largest proportion was observed in `days_since_maintenance`
(5.24%), followed by `downtime_hours_30d` (3.90%) and `vibration_mm_s`
(3.86%).

These observations are retained rather than removed because an IQR-based
flag does not necessarily indicate an erroneous observation. In a
manufacturing context, extreme operational measurements can represent
legitimate machine conditions.

Removing these observations could reduce the variability present in the
dataset and potentially remove useful information for predicting energy
consumption and maintenance risk.

Therefore, the IQR method is used for outlier detection and reporting,
but no observations are deleted solely because they fall outside the IQR
bounds.


 1. Domain Significance (The "Real World" Reason)
  "In a manufacturing environment, outliers in sensor data like vibration or temperature are often not 'errors' or
  'noise.' They represent extreme operational states. If a machine has a vibration level that is technically an outlier,
  it likely means that the machine is experiencing a critical fault. Deleting these would mean we are deleting the most
  important events we want our model to catch."

  2. Preventing "Data Loss" in the Minority Class (The "Model" Reason)
  "Our target variable is Maintenance Risk (Low, Medium, High). The 'High Risk' machines are the minority in our
  dataset. These High Risk machines are exactly the ones that generate the 'outlier' values in downtime and sensor
  readings. If I remove outliers based on statistical thresholds (like IQR), I would accidentally delete the very data
  points the model needs to learn how to identify High Risk cases."

  3. Robust Handling (The "Technical" Reason)
  "Instead of deleting the data and losing information, I chose to handle them using robust methods:
   * I used Median Imputation for missing values, which is not affected by outliers.
   * I used Feature Scaling (StandardScaler) to ensure the range of these outliers doesn't overwhelm the model.
   * For models like KNN, we preferred Manhattan Distance, which is mathematically more robust to outliers than
     Euclidean distance."

## B3.7 — Architectural Rationale: Why Label Encoding over One-Hot Encoding

1. **Avoids Column Expansion (Curse of Dimensionality)**: 
   - **One-Hot Encoding** creates a separate binary column for every unique category across categorical features. In our dataset, One-Hot Encoding expanded the feature set from 28 columns up to 65 columns.
   - **Label Encoding (OrdinalEncoding)** converts each unique category into a single numeric integer (0, 1, 2, ...), keeping the number of predictor columns unchanged at **28 features**.

2. **Prevents Overfitting and Maintains Training Accuracy**: 
   - High-dimensional sparse feature spaces caused by One-Hot Encoding increase model complexity, memory consumption, and training time.
   - Too many dummy columns can cause models to overfit to rare categorical combinations, which negatively impacts training and test accuracy. Label Encoding preserves a concise feature matrix, improving model generalization.

3. **Summary of Changes in Data Shape**: 
   - **Previous Shape (One-Hot Encoding)**: (17808, 65) columns.
   - **New Shape (Label Encoding)**: (17808, 28) columns.

````python
# =====================================================================
# Transition from Baseline to Best-Fit Line (Gradient Descent)
# =====================================================================
# Visualizes the step-by-step transition of Linear Regression from an
# initial flat baseline (zero knowledge) into the optimal best-fit line.

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# 1. Select the primary physical predictor (motor_current_a vs machine_energy_kwh)
if "df" in globals() or "df" in locals():
    data_source = df
else:
    import pandas as pd
    data_source = pd.read_csv("smart_manufacturing_ml_dataset.csv")

# Sample points for a clean, publication-quality scatter visual
sample_pts = data_source.sample(n=350, random_state=42).sort_values(by="motor_current_a")
x_raw = sample_pts["motor_current_a"].values.reshape(-1, 1)
y_raw = sample_pts["machine_energy_kwh"].values.reshape(-1, 1)

# Standardize features for smooth, stable Gradient Descent convergence
x_mean, x_std = x_raw.mean(), x_raw.std()
y_mean, y_std = y_raw.mean(), y_raw.std()

x_norm = (x_raw - x_mean) / x_std
y_norm = (y_raw - y_mean) / y_std

m_samples = len(x_norm)
X_b = np.c_[np.ones((m_samples, 1)), x_norm]

# 2. Gradient Descent Optimization: Track progression from Baseline to Best-Fit Line
params = np.array([[0.0], [0.0]])  # Baseline: flat line at mean(y)
learning_rate = 0.1
n_iterations = 40
save_every = 2
lines_history = []

for iteration in range(n_iterations):
    # Convert standardized parameters back to original data coordinates
    b_norm, m_norm = params[0, 0], params[1, 0]
    m_orig = m_norm * (y_std / x_std)
    b_orig = y_mean + y_std * b_norm - m_orig * x_mean

    if iteration % save_every == 0 or iteration == n_iterations - 1:
        lines_history.append((iteration, m_orig, b_orig))

    # Compute gradients and update parameters
    gradients = (2 / m_samples) * X_b.T.dot(X_b.dot(params) - y_norm)
    params = params - learning_rate * gradients

# 3. Plot the Transition in One Single Graph
plt.figure(figsize=(11, 6), dpi=120)

# Scatter plot of actual observations
plt.scatter(
    x_raw, y_raw,
    color="#2b5c8f", alpha=0.45, edgecolors="none", s=35,
    label="Observed Data (Motor Current vs Energy)"
)

# Colormap for transitions (cool to warm progression)
colormap = cm.viridis
x_line = np.linspace(x_raw.min(), x_raw.max(), 150)

# Plot intermediate transition lines
for idx, (epoch, slope, intercept) in enumerate(lines_history):
    color = colormap(idx / len(lines_history))
    y_line = slope * x_line + intercept

    if idx == 0:
        # Baseline line (flat initial prediction)
        plt.plot(
            x_line, y_line,
            color="#555555", linestyle="--", linewidth=2.4,
            label=f"Baseline (Epoch 0: y = {intercept:.2f})"
        )
    elif idx == len(lines_history) - 1:
        # Final Converged Best-Fit line
        plt.plot(
            x_line, y_line,
            color="#e63946", linewidth=3.2,
            label=f"Best Fit Line (Final: y = {slope:.2f}x + {intercept:.2f})"
        )
    else:
        # Intermediate transitions
        plt.plot(x_line, y_line, color=color, alpha=0.35, linewidth=1.3)

# 4. Graph Styling & Annotations
plt.title("Linear Regression: Transition from Baseline to Best-Fit Line", fontsize=13, weight="bold", pad=12)
plt.xlabel("Motor Current (A)", fontsize=11)
plt.ylabel("Machine Energy Consumption (kWh)", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.5)

# Colorbar showing iteration progression
sm = plt.cm.ScalarMappable(cmap=colormap, norm=plt.Normalize(vmin=0, vmax=n_iterations))
sm.set_array([])
cbar = plt.colorbar(sm, ax=plt.gca(), pad=0.02)
cbar.set_label("Training Iterations / Epochs (Gradient Descent)", fontsize=10)

plt.legend(frameon=True, facecolor="white", edgecolor="#cccccc", loc="upper left", fontsize=9.5)
plt.tight_layout()
plt.show()
````

Polynomial Regression extends Linear Regression by creating polynomial and interaction features from the existing predictors. This allows the model to learn non-linear relationships between variables.

Since the processed dataset contains 28 features, using high polynomial degrees would create an extremely large feature space, increasing computational cost and the risk of overfitting. Therefore, polynomial degrees will be compared incrementally while keeping the same train-test split.

````python
#This allows the model to learn non-linear relationships between variables.

from sklearn.preprocessing import PolynomialFeatures

degrees = [1, 2]

poly_results = []
poly_models = {}
poly_transformers = {}

for degree in degrees:
    poly = PolynomialFeatures(degree=degree, include_bias=False)
    
    X_train_poly = poly.fit_transform(X_reg_train_processed)
    X_test_poly = poly.transform(X_reg_test_processed)
    
    model = LinearRegression()
    model.fit(X_train_poly, y_reg_train)
    
    y_pred = model.predict(X_test_poly)
    
    r2 = r2_score(y_reg_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_reg_test, y_pred))
    mae = mean_absolute_error(y_reg_test, y_pred)
    
    poly_results.append({
        "Degree": degree,
        "Polynomial Features": X_train_poly.shape[1],
        "R²": r2,
        "RMSE": rmse,
        "MAE": mae
    })
    
    poly_models[degree] = model
    poly_transformers[degree] = poly

poly_results = pd.DataFrame(poly_results)

poly_results
````

````python
# =====================================================================
# Gradient Boosting: Staged Loss Reduction across Boosting Iterations
# =====================================================================
# Visualizes how sequential boosting iteratively reduces test and train
# MSE deviance across 100 consecutive estimators.

import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import mean_squared_error

# Compute MSE at each boosting stage
train_deviance = np.zeros((gradient_boosting_model.n_estimators,), dtype=np.float64)
test_deviance = np.zeros((gradient_boosting_model.n_estimators,), dtype=np.float64)

for i, y_pred in enumerate(gradient_boosting_model.staged_predict(X_reg_train_processed)):
    train_deviance[i] = mean_squared_error(y_reg_train, y_pred)

for i, y_pred in enumerate(gradient_boosting_model.staged_predict(X_reg_test_processed)):
    test_deviance[i] = mean_squared_error(y_reg_test, y_pred)

plt.figure(figsize=(10, 5.5), dpi=120)
iterations = np.arange(1, gradient_boosting_model.n_estimators + 1)

plt.plot(iterations, train_deviance, color="#2b5c8f", linewidth=2.4, label="Training Deviance (MSE)")
plt.plot(iterations, test_deviance, color="#e63946", linewidth=2.4, linestyle="--", label="Test Deviance (MSE)")

best_stage = np.argmin(test_deviance) + 1
plt.axvline(best_stage, color="#06d6a0", linestyle=":", linewidth=2, label=f"Optimal Estimators: {best_stage}")

plt.title("Gradient Boosting: Sequential Loss Reduction (Staged Deviance)", fontsize=12, weight="bold", pad=12)
plt.xlabel("Boosting Iterations (Number of Trees)", fontsize=11)
plt.ylabel("Mean Squared Error (Deviance)", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.5)
plt.legend(frameon=True, facecolor="white", edgecolor="#cccccc", loc="upper right")
plt.tight_layout()
plt.show()
````

## C1.9 — Support Vector Regressor (SVR)

Support Vector Regressor (SVR) maps input features into a high-dimensional space using kernel functions to fit a hyper-plane within an epsilon margin of tolerance. Features are standardized first using StandardScaler, and hyperparameters C and kernel are tuned using GridSearchCV.

````python
from sklearn.svm import SVR
from sklearn.model_selection import GridSearchCV

# SVR Hyperparameter Tuning using GridSearchCV on scaled features
svr_param_grid = {
    "C": [0.1, 1.0, 10.0],
    "kernel": ["rbf"]
}

svr_grid = GridSearchCV(
    estimator=SVR(),
    param_grid=svr_param_grid,
    scoring="r2",
    cv=3,
    n_jobs=-1
)

svr_grid.fit(X_reg_train_processed[:4000], y_reg_train[:4000])

# Fit best SVR model on full training set
best_svr_model = SVR(C=svr_grid.best_params_["C"], kernel=svr_grid.best_params_["kernel"])
best_svr_model.fit(X_reg_train_processed, y_reg_train)

# Evaluate SVR predictions
y_pred_svr = best_svr_model.predict(X_reg_test_processed)

svr_r2 = r2_score(y_reg_test, y_pred_svr)
svr_rmse = np.sqrt(mean_squared_error(y_reg_test, y_pred_svr))
svr_mae = mean_absolute_error(y_reg_test, y_pred_svr)

print("Best SVR Parameters:", svr_grid.best_params_)
print("Support Vector Regressor (SVR) Results")
print("---------------------------------------")
print(f"R²   : {svr_r2:.4f}")
print(f"RMSE : {svr_rmse:.4f}")
print(f"MAE  : {svr_mae:.4f}")
````

````python
# =====================================================================
# Support Vector Regressor: Epsilon-Insensitive Tube & Support Vectors
# =====================================================================
# Demonstrates SVR fitting within an epsilon margin tolerance tube (y ± ε),
# highlighting the critical Support Vectors that determine the boundary.

import matplotlib.pyplot as plt
import numpy as np
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler

sample_svr = df.sample(n=250, random_state=42).sort_values(by="motor_current_a")
x_raw_svr = sample_svr["motor_current_a"].values.reshape(-1, 1)
y_raw_svr = sample_svr["machine_energy_kwh"].values

scaler_x = StandardScaler()
scaler_y = StandardScaler()
x_scaled = scaler_x.fit_transform(x_raw_svr)
y_scaled = scaler_y.fit_transform(y_raw_svr.reshape(-1, 1)).ravel()

# Fit SVR with visible epsilon margin
epsilon_val = 0.25
svr_demo = SVR(kernel="rbf", C=10.0, epsilon=epsilon_val)
svr_demo.fit(x_scaled, y_scaled)

# Support vectors
sv_indices = svr_demo.support_
x_dense_s = np.linspace(x_scaled.min(), x_scaled.max(), 300).reshape(-1, 1)
y_dense_s = svr_demo.predict(x_dense_s)

# Map back to original physical units
x_dense_orig = scaler_x.inverse_transform(x_dense_s).ravel()
y_dense_orig = scaler_y.inverse_transform(y_dense_s.reshape(-1, 1)).ravel()
epsilon_orig = epsilon_val * scaler_y.scale_[0]

plt.figure(figsize=(10, 6), dpi=120)
# Regular points
plt.scatter(x_raw_svr, y_raw_svr, color="#2b5c8f", alpha=0.5, s=35, label="Within Margin Points (Zero Loss)")
# Highlight Support Vectors
plt.scatter(x_raw_svr[sv_indices], y_raw_svr[sv_indices], facecolors="none", edgecolors="#e63946", s=90, linewidth=1.8, label=f"Support Vectors ({len(sv_indices)} vectors)")

# Fitted line and epsilon tube
plt.plot(x_dense_orig, y_dense_orig, color="#1d3557", linewidth=2.8, label="SVR RBF Prediction")
plt.fill_between(x_dense_orig, y_dense_orig - epsilon_orig, y_dense_orig + epsilon_orig, color="#a8dadc", alpha=0.4, label=f"Epsilon Tube (±{epsilon_orig:.2f} kWh)")

plt.title("Support Vector Regressor: Epsilon-Insensitive Margin Tube & Support Vectors", fontsize=12, weight="bold", pad=12)
plt.xlabel("Motor Current (A)", fontsize=11)
plt.ylabel("Machine Energy Consumption (kWh)", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.5)
plt.legend(frameon=True, facecolor="white", edgecolor="#cccccc", loc="upper left")
plt.tight_layout()
plt.show()
````

### Classification Evaluation Helper Function

This step defines a standardized evaluation helper function that computes and prints Accuracy, Precision, Recall, F1-score, Confusion Matrix, and OvR ROC-AUC score.


````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_auc_score
)

# Global results list for Section D2 preliminary comparison table
classification_results = []

def evaluate_classifier(model_name, y_true, y_pred, y_prob):
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average="weighted", zero_division=0)
    rec = recall_score(y_true, y_pred, average="weighted", zero_division=0)
    f1 = f1_score(y_true, y_pred, average="weighted", zero_division=0)
    roc_auc = roc_auc_score(y_true, y_prob, multi_class="ovr", average="weighted")
    cm = confusion_matrix(y_true, y_pred)
    
    classification_results.append({
        "Algorithm": model_name,
        "Accuracy": round(acc, 4),
        "Precision": round(prec, 4),
        "Recall": round(rec, 4),
        "F1-score (Weighted)": round(f1, 4),
        "ROC-AUC (OvR)": round(roc_auc, 4)
    })
    
    print(f"=== {model_name} Evaluation ===")
    print(f"Accuracy            : {acc:.4f}")
    print(f"Precision (Weighted): {prec:.4f}")
    print(f"Recall (Weighted)   : {rec:.4f}")
    print(f"F1-score (Weighted) : {f1:.4f}")
    print(f"ROC-AUC (OvR)       : {roc_auc:.4f}")
    print("\nConfusion Matrix:")
    print(cm)
    
    # Plot Confusion Matrix
    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=label_encoder.classes_
    )
    fig, ax = plt.subplots(figsize=(6, 5))
    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    plt.title(f"Confusion Matrix — {model_name}")
    plt.tight_layout()
    plt.show()
````

## D1.1 — Logistic Regression (Baseline Classifier)

Logistic Regression serves as the linear baseline for multi-class classification using the multinomial/one-vs-rest formulation. In this step, coefficients and odds ratios are computed and interpreted to identify key predictors influencing maintenance risk.


````python
from sklearn.linear_model import LogisticRegression

# 1. Train Multinomial Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_class_train_processed, y_class_train)

# Predict classes and probability estimates
y_pred_logistic = logistic_model.predict(X_class_test_processed)
y_prob_logistic = logistic_model.predict_proba(X_class_test_processed)

# Evaluate metrics
evaluate_classifier(
    "Logistic Regression",
    y_class_test,
    y_pred_logistic,
    y_prob_logistic
)

# Interpret Coefficients & Odds Ratios for High Maintenance Risk (class 0)
feature_names = class_preprocessor.get_feature_names_out()
high_risk_coef = logistic_model.coef_[0]
odds_ratios = np.exp(high_risk_coef)

coef_df = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": high_risk_coef,
    "Odds_Ratio": odds_ratios
}).sort_values(by="Odds_Ratio", ascending=False)

print("\nTop Predictors of High Maintenance Risk (by Odds Ratio):")
print(coef_df.head(10).to_string(index=False))
````

### Logistic Regression Interpretation

**Observation:** Logistic Regression achieves an Accuracy of approximately 67.3% and an OvR ROC-AUC score of 0.8249. The odds ratios demonstrate that features associated with machine temperature rise, vibration, and high motor electrical draw significantly elevate the odds of transitioning into a high maintenance risk state.

````python
# ============================================================
# COMBINED MULTICLASS ROC CURVE + AUC FOR ALL 5 MODELS
# ============================================================

from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt
import numpy as np

# Your 5 classes
classes = np.unique(y_class_test)

# Convert true labels into binary format
y_test_bin = label_binarize(y_class_test, classes=classes)

# Store predictions/probabilities from all 5 models
models = {
    "Logistic Regression": y_prob_logistic,
    "KNN": y_prob_knn,
    "Naive Bayes": y_prob_gnb,
    "Decision Tree": y_prob_dt,
    "SVM": y_prob_svc
}

# Store AUC values
auc_results = {}

# Create combined ROC plot
plt.figure(figsize=(10, 7))

for model_name, y_prob in models.items():

    # Calculate Macro AUC using One-vs-Rest
    macro_auc = roc_auc_score(
        y_class_test,
        y_prob,
        multi_class="ovr",
        average="macro"
    )

    auc_results[model_name] = macro_auc

    # Calculate ROC curve for each class
    fpr = {}
    tpr = {}

    for i in range(len(classes)):
        fpr[i], tpr[i], _ = roc_curve(
            y_test_bin[:, i],
            y_prob[:, i]
        )

    # Create macro-average ROC curve
    all_fpr = np.unique(
        np.concatenate([
            fpr[i] for i in range(len(classes))
        ])
    )

    mean_tpr = np.zeros_like(all_fpr)

    for i in range(len(classes)):
        mean_tpr += np.interp(
            all_fpr,
            fpr[i],
            tpr[i]
        )

    mean_tpr /= len(classes)

    # Plot model's macro-average ROC
    plt.plot(
        all_fpr,
        mean_tpr,
        linewidth=2,
        label=f"{model_name} (AUC = {macro_auc:.3f})"
    )


# Random classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random Classifier"
)

# Graph labels
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)

plt.title(
    "Combined Multiclass ROC Curve - Model Comparison",
    fontsize=14
)

plt.legend(loc="lower right")
plt.grid(True)
plt.tight_layout()
plt.show()


# ============================================================
# PRINT AUC FOR EACH MODEL
# ============================================================

print("\n" + "=" * 50)
print("AUC RESULTS - MULTICLASS ROC")
print("=" * 50)

for model_name, auc_value in auc_results.items():
    print(f"{model_name:<25} : {auc_value:.4f}")

print("=" * 50)
````

### E0.1 — Reusable Tune-and-Evaluate Helper (Part B)

To keep the notebook modular, one helper handles every Part-B algorithm:

1. **Baseline** – fit the model with its starting hyper-parameters (training split only) and record test Accuracy / weighted F1 / ROC-AUC.
2. **Tuning** – `GridSearchCV` (3-fold stratified CV on the training split, scoring = weighted F1).
3. **Evaluation** – the best estimator is evaluated on the shared test split with the existing `evaluate_classifier()` (Accuracy, Precision, Recall, weighted F1, ROC-AUC OvR, confusion matrix), which also appends the result to `classification_results`.
4. **Logging** – the default-vs-tuned improvement is stored for the Part-B tuning summary.

````python
part_b_tuning_log = []   # default-vs-tuned record for each Part-B model


def tune_and_evaluate(name, base_estimator, param_grid, cv=CV3):
    """Baseline fit -> GridSearchCV -> test evaluation -> logging. Returns (best_estimator, fitted_search)."""
    start = time.time()

    # 1) Baseline with the starting hyper-parameters (trained on the training split only)
    baseline = clone(base_estimator).fit(X_class_train_processed, y_class_train)
    base_pred = baseline.predict(X_class_test_processed)
    base_prob = baseline.predict_proba(X_class_test_processed)
    base = {
        "Accuracy": accuracy_score(y_class_test, base_pred),
        "F1":       f1_score(y_class_test, base_pred, average="weighted"),
        "ROC-AUC":  roc_auc_score(y_class_test, base_prob, multi_class="ovr", average="weighted"),
    }

    # 2) Hyper-parameter tuning on the training split only
    search = GridSearchCV(
        clone(base_estimator), param_grid,
        scoring="f1_weighted", cv=cv, n_jobs=-1
    )
    search.fit(X_class_train_processed, y_class_train)
    best_model = search.best_estimator_

    # 3) Evaluate the tuned model on the shared test split
    y_pred = best_model.predict(X_class_test_processed)
    y_prob = best_model.predict_proba(X_class_test_processed)
    tuned = {
        "Accuracy": accuracy_score(y_class_test, y_pred),
        "F1":       f1_score(y_class_test, y_pred, average="weighted"),
        "ROC-AUC":  roc_auc_score(y_class_test, y_prob, multi_class="ovr", average="weighted"),
    }

    print(f"##### {name} #####")
    print("Best parameters      :", search.best_params_)
    print(f"Best CV weighted F1  : {search.best_score_:.4f}")
    print(f"Default test F1      : {base['F1']:.4f}")
    print(f"Tuned test F1        : {tuned['F1']:.4f}  (change: {tuned['F1'] - base['F1']:+.4f})")
    print(f"Search time          : {time.time() - start:.1f} s\n")

    top5 = (
        pd.DataFrame(search.cv_results_)
        [["params", "mean_test_score", "std_test_score", "rank_test_score"]]
        .sort_values("rank_test_score").head(5)
        .rename(columns={"params": "Parameters", "mean_test_score": "Mean CV F1",
                         "std_test_score": "Std CV F1", "rank_test_score": "Rank"})
    )
    top5["Parameters"] = top5["Parameters"].astype(str)
    display(top5.reset_index(drop=True))

    # Standard metrics, confusion matrix and appending to classification_results
    evaluate_classifier(name, y_class_test, y_pred, y_prob)

    # Register model + probabilities and log the improvement (replace entry if the cell is re-run)
    trained_classifiers[name] = best_model
    class_probabilities[name] = y_prob
    part_b_tuning_log[:] = [r for r in part_b_tuning_log if r["Algorithm"] != name]
    part_b_tuning_log.append({
        "Algorithm": name,
        "Default Accuracy": base["Accuracy"], "Tuned Accuracy": tuned["Accuracy"],
        "Default F1": base["F1"],             "Tuned F1": tuned["F1"],
        "Default ROC-AUC": base["ROC-AUC"],   "Tuned ROC-AUC": tuned["ROC-AUC"],
        "CV F1 (best)": search.best_score_,
        "Best Parameters": str(search.best_params_),
    })
    return best_model, search
````

### E7.2 — Combined Multi-Class ROC Curves (All 10 Algorithms, One-vs-Rest)

**Why we plot this:** ROC curves show the true-positive / false-positive trade-off across all decision thresholds. For the three-class problem, one One-vs-Rest curve is computed per class and averaged (macro average). The legend reports the macro-average AUC, which can differ slightly from the *weighted* ROC-AUC in the table because the table weights classes by their frequency.

````python
def plot_macro_roc(prob_by_model, y_true, title):
    """Macro-averaged One-vs-Rest ROC curve for every model in prob_by_model."""
    class_ids = np.unique(y_true)
    y_bin = label_binarize(y_true, classes=class_ids)
    fig, ax = plt.subplots(figsize=(11, 8))
    macro_aucs = {}

    for color, (name, prob) in zip(PALETTE, prob_by_model.items()):
        macro_aucs[name] = roc_auc_score(y_true, prob, multi_class="ovr", average="macro")
        fpr, tpr = {}, {}
        for i in range(len(class_ids)):
            fpr[i], tpr[i], _ = roc_curve(y_bin[:, i], prob[:, i])
        grid_fpr = np.unique(np.concatenate([fpr[i] for i in range(len(class_ids))]))
        mean_tpr = np.mean([np.interp(grid_fpr, fpr[i], tpr[i]) for i in range(len(class_ids))], axis=0)
        ax.plot(grid_fpr, mean_tpr, color=color, lw=2, label=f"{name} (AUC = {macro_aucs[name]:.3f})")

    ax.plot([0, 1], [0, 1], "k--", lw=1.2, label="Random classifier")
    ax.set_title(title, fontsize=14, weight="bold")
    ax.set_xlabel("False Positive Rate")
    ax.set_ylabel("True Positive Rate")
    ax.legend(loc="lower right", fontsize=9)
    ax.grid(True, linestyle=":", alpha=0.6)
    plt.tight_layout()
    plt.show()
    return macro_aucs


ordered_probs = {name: class_probabilities[name] for name in comparison_df["Algorithm"]}
macro_auc_all = plot_macro_roc(ordered_probs, y_class_test, "Macro-Average One-vs-Rest ROC Curves - All 10 Classifiers")
````

## E8 — Final Model Selection, Cross-Validation & Hyper-parameter Tuning Evidence

The final model is chosen with a transparent rule: **highest weighted F1-score on the shared test split (ROC-AUC as tie-breaker)**. To confirm the ranking is not an artefact of a single split, **5-fold stratified cross-validation** is run on the training data for the top two models. Finally, the tuning evidence for the selected model is shown as *before vs. after* metrics:

* if the selected model is a **Part-B** model, its default-vs-tuned results from E6 are used;
* if it is a **Part-A** model that was not tuned in Review 1, a `GridSearchCV` is run here (5-fold) and compared against its Review-1 result.

````python
# ------------------------------------------------------------
# 1) Identify the top two models from the consolidated table
# ------------------------------------------------------------
best_name = comparison_df.loc[0, "Algorithm"]
runner_up_name = comparison_df.loc[1, "Algorithm"]
print("Selected (rank 1):", best_name)
print("Runner-up (rank 2):", runner_up_name)


# ------------------------------------------------------------
# 2) 5-fold stratified CV on the TRAINING split for the top two models
# ------------------------------------------------------------
def five_fold_summary(name):
    estimator = clone(trained_classifiers[name])
    if "probability" in estimator.get_params():
        estimator.set_params(probability=False)      # not needed for CV scoring; much faster for SVC
    cv5 = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    scores = cross_validate(
        estimator, X_class_train_processed, y_class_train, cv=cv5,
        scoring={"accuracy": "accuracy", "f1": "f1_weighted"}, n_jobs=-1
    )
    return {
        "Algorithm": name,
        "CV Accuracy (mean)": scores["test_accuracy"].mean(),
        "CV Accuracy (std)": scores["test_accuracy"].std(),
        "CV Weighted F1 (mean)": scores["test_f1"].mean(),
        "CV Weighted F1 (std)": scores["test_f1"].std(),
    }


cv_top2 = pd.DataFrame([five_fold_summary(best_name), five_fold_summary(runner_up_name)])
print("\n5-fold stratified cross-validation (top two models):")
display(cv_top2.round(4))
cv_top2.to_csv(f"{RESULTS_DIR}/classification_top2_cv.csv", index=False)
````

````python
# ------------------------------------------------------------
# 3) Hyper-parameter tuning evidence for the selected model
# ------------------------------------------------------------
# Search spaces for Part-A models (used only if one of them is selected as the final model)
part_a_specs = {
    "Logistic Regression": (
        LogisticRegression(max_iter=2000, random_state=RANDOM_STATE),
        {"C": [0.01, 0.1, 1, 10, 100], "class_weight": [None, "balanced"]}),
    "K-Nearest Neighbors (KNN)": (
        KNeighborsClassifier(weights="distance"),
        {"n_neighbors": [5, 9, 13, 17, 25], "metric": ["euclidean", "manhattan"]}),
    "Naive Bayes (Gaussian)": (
        GaussianNB(),
        {"var_smoothing": np.logspace(-11, -5, 7)}),
    "Decision Tree Classifier": (
        DecisionTreeClassifier(random_state=RANDOM_STATE),
        {"max_depth": [3, 4, 5, 6, 8], "min_samples_leaf": [5, 20, 50]}),
    "Support Vector Machine (SVC)": (
        SVC(kernel="rbf", probability=True, random_state=RANDOM_STATE),
        {"C": [0.5, 1, 5, 10], "gamma": ["scale", 0.01]}),
}


def metric_triplet(model):
    pred = model.predict(X_class_test_processed)
    prob = model.predict_proba(X_class_test_processed)
    return {
        "Accuracy": accuracy_score(y_class_test, pred),
        "Weighted F1": f1_score(y_class_test, pred, average="weighted"),
        "ROC-AUC (OvR)": roc_auc_score(y_class_test, prob, multi_class="ovr", average="weighted"),
    }


if best_name in PART_B_NAMES:
    # Tuning already carried out in E1-E5 -> reuse the logged default-vs-tuned results
    log = part_b_tuning_df.set_index("Algorithm").loc[best_name]
    before = {"Accuracy": log["Default Accuracy"], "Weighted F1": log["Default F1"], "ROC-AUC (OvR)": log["Default ROC-AUC"]}
    after = {"Accuracy": log["Tuned Accuracy"], "Weighted F1": log["Tuned F1"], "ROC-AUC (OvR)": log["Tuned ROC-AUC"]}
    final_model = trained_classifiers[best_name]
    final_params = log["Best Parameters"]
    before_label = "Default hyper-parameters"
else:
    # Part-A model that was not tuned in Review 1 -> run a 5-fold grid search now
    base_estimator, grid = part_a_specs[best_name]
    search_estimator = clone(base_estimator)
    if "probability" in search_estimator.get_params():
        search_estimator.set_params(probability=False)   # speeds up the search; restored below
    final_search = GridSearchCV(
        search_estimator, grid, scoring="f1_weighted",
        cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE),
        n_jobs=-1, refit=False
    )
    final_search.fit(X_class_train_processed, y_class_train)
    final_model = clone(base_estimator).set_params(**final_search.best_params_)
    final_model.fit(X_class_train_processed, y_class_train)
    row = comparison_df.loc[0]
    before = {"Accuracy": row["Accuracy"], "Weighted F1": row["F1-score (Weighted)"], "ROC-AUC (OvR)": row["ROC-AUC (OvR)"]}
    after = metric_triplet(final_model)
    final_params = str(final_search.best_params_)
    before_label = "Review-1 configuration"
    trained_classifiers[best_name] = final_model

tuning_effect_df = pd.DataFrame({
    "Metric": list(before.keys()),
    before_label: [before[k] for k in before],
    "After tuning": [after[k] for k in before],
})
tuning_effect_df["Change"] = tuning_effect_df["After tuning"] - tuning_effect_df[before_label]

print(f"Final model: {best_name}")
print(f"Best hyper-parameters: {final_params}\n")
display(tuning_effect_df.round(4))
tuning_effect_df.round(4).to_csv(f"{RESULTS_DIR}/final_model_tuning_effect.csv", index=False)

MEANINGFUL_CHANGE = 1e-4
gains = tuning_effect_df[tuning_effect_df["Change"] > MEANINGFUL_CHANGE]["Metric"].tolist()
if gains:
    print("Tuning improved:", ", ".join(gains))
else:
    print("Tuning produced no meaningful change for this model (the starting configuration was already near-optimal).")
    print("Default-vs-tuned improvements for all five Part-B models are documented in the E6 table above.")
````

# Section F — Clustering Track (K-Means & Agglomerative Hierarchical Clustering)

Clustering is **unsupervised**: the algorithms never see the target. The ground-truth `maintenance_risk` labels are used **only after fitting**, to validate and interpret the discovered clusters. The regression target (`machine_energy_kwh`), `machine_id` and `maintenance_risk` are all excluded from the clustering features.

| Step | Content |
|---|---|
| F1 | Feature selection, sampling and scaling |
| F2 | K-Means: Elbow curve → choice of *k* → fit |
| F3 | Agglomerative clustering: linkage comparison + **Dendrograms** |
| F4 | Evaluation: Silhouette, Davies-Bouldin, Calinski-Harabasz for both algorithms |
| F5 | **PCA 2-D** cluster plots and **t-SNE** plot |
| F6 | Cluster profiling and post-hoc validation against `maintenance_risk` |

## F1 — Clustering Data Preparation

* **Features:** the continuous numerical sensor/operational variables plus the engineered `bearing_temp_rise_c`. Ordinal-encoded categorical columns are excluded because integer codes have no meaningful distance, which would distort Euclidean-distance clustering.
* **Outlier handling:** extreme values are winsorised (clipped to the 1st-99th percentile of each feature) before scaling. Distance-based clustering is very sensitive to outliers: without this step a few abnormal machines form their own tiny clusters instead of revealing the main operating groups.
* **Scaling:** `StandardScaler`, so features with large ranges (e.g. `rpm`, `batch_size`) do not dominate the distance. There is no train/test split in unsupervised learning, so the scaler is fitted on the clustering sample itself.
* **Sampling:** Agglomerative clustering needs memory quadratic in the number of rows. A reproducible random sample (`random_state=42`) is therefore used for **all** clustering algorithms and metrics, so every comparison is on identical data.

````python
import sys
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score,
    normalized_mutual_info_score,
)
from scipy.cluster.hierarchy import linkage, dendrogram, cophenet, set_link_color_palette
from scipy.spatial.distance import pdist

CLUSTER_SAMPLE_SIZE = 4000

# Features: numeric predictors only (no IDs, no targets)
exclude_from_clustering = {"machine_id", regression_target, classification_target}
cluster_features = [c for c in numerical_features if c not in exclude_from_clustering]
if "bearing_temp_rise_c" not in cluster_features:
    cluster_features.append("bearing_temp_rise_c")

# Reproducible sample; the risk labels are set aside and used ONLY for post-hoc validation
cluster_df = df_clean.sample(n=min(CLUSTER_SAMPLE_SIZE, len(df_clean)), random_state=RANDOM_STATE)
risk_labels_for_validation = cluster_df[classification_target].copy()

# Winsorise: clip each feature to its 1st-99th percentile so a handful of extreme values
# (e.g. a few machines with abnormal downtime) cannot dominate the distance calculations
lower_clip = cluster_df[cluster_features].quantile(0.01)
upper_clip = cluster_df[cluster_features].quantile(0.99)
cluster_clean = cluster_df[cluster_features].clip(lower=lower_clip, upper=upper_clip, axis=1)
n_clipped = int(((cluster_df[cluster_features] < lower_clip) | (cluster_df[cluster_features] > upper_clip)).sum().sum())

cluster_scaler = StandardScaler()
X_cluster = cluster_scaler.fit_transform(cluster_clean)

print("Clustering features:", len(cluster_features))
print(cluster_features)
print("\nClustering matrix shape:", X_cluster.shape)
print("Missing values         :", int(np.isnan(X_cluster).sum()))
print(f"Values winsorised (1st-99th pct): {n_clipped:,} of {X_cluster.size:,} ({n_clipped / X_cluster.size:.2%})")
print("Mean / std after scaling:", round(float(X_cluster.mean()), 4), "/", round(float(X_cluster.std()), 4))
````

### F1.1 — Reusable Clustering Helpers

Two small functions keep the clustering code modular: one computes the three mandatory internal validity metrics, the other finds the "elbow" of an inertia curve geometrically (the point farthest from the straight line joining the first and last points, after normalising both axes).

| Metric | Direction | Meaning |
|---|---|---|
| Silhouette Score | higher is better (−1 … 1) | how much closer a point is to its own cluster than to the nearest other cluster |
| Davies-Bouldin Index | **lower** is better (≥ 0) | average similarity between each cluster and its most similar cluster |
| Calinski-Harabasz Index | higher is better | ratio of between-cluster to within-cluster dispersion |

````python
def evaluate_clustering(X, labels):
    """Mandatory internal validity metrics for one clustering result."""
    return {
        "Silhouette Score": silhouette_score(X, labels),
        "Davies-Bouldin Index": davies_bouldin_score(X, labels),
        "Calinski-Harabasz Index": calinski_harabasz_score(X, labels),
    }


def find_elbow(k_values, inertias):
    """Geometric elbow: point with the largest distance to the line joining the curve's end points."""
    pts = np.column_stack([k_values, inertias]).astype(float)
    pts = (pts - pts.min(axis=0)) / (pts.max(axis=0) - pts.min(axis=0))
    direction = pts[-1] - pts[0]
    direction = direction / np.linalg.norm(direction)
    offsets = pts - pts[0]
    distances = np.abs(offsets[:, 0] * direction[1] - offsets[:, 1] * direction[0])
    return int(np.asarray(k_values)[np.argmax(distances)])


def scatter_clusters(ax, coords, labels, title, xlabel, ylabel, centers=None):
    """2-D scatter with one colour-blind-friendly colour per cluster and a legend."""
    cluster_ids = np.unique(labels)
    colors = sns.color_palette("colorblind", n_colors=len(cluster_ids))
    for color, cid in zip(colors, cluster_ids):
        mask = labels == cid
        ax.scatter(coords[mask, 0], coords[mask, 1], s=12, alpha=0.6, color=color,
                   label=f"Cluster {cid} (n={mask.sum():,})")
    if centers is not None:
        ax.scatter(centers[:, 0], centers[:, 1], marker="X", s=220, c="black",
                   edgecolors="white", linewidths=1.5, label="Centroids")
    ax.set_title(title, fontsize=12, weight="bold")
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.legend(fontsize=8, loc="best")
````

## F2 — K-Means Clustering & the Elbow Curve

K-Means partitions the data into *k* clusters by repeatedly assigning points to the nearest centroid and recomputing centroids. Because *k* must be chosen in advance, K-Means is fitted for k = 2 … 10 and **inertia** (within-cluster sum of squared distances) is plotted against *k*. The "elbow" is where adding more clusters stops reducing inertia sharply. The Silhouette Score per *k* is shown next to it as a cross-check.

````python
K_RANGE = list(range(2, 11))
scan_rows = []
for k in K_RANGE:
    km = KMeans(n_clusters=k, n_init=10, random_state=RANDOM_STATE).fit(X_cluster)
    scan_rows.append({
        "k": k,
        "Inertia": km.inertia_,
        "Silhouette": silhouette_score(X_cluster, km.labels_),
        "Davies-Bouldin": davies_bouldin_score(X_cluster, km.labels_),
        "Calinski-Harabasz": calinski_harabasz_score(X_cluster, km.labels_),
    })
kmeans_scan = pd.DataFrame(scan_rows)

elbow_k = find_elbow(kmeans_scan["k"], kmeans_scan["Inertia"])
silhouette_best_k = int(kmeans_scan.loc[kmeans_scan["Silhouette"].idxmax(), "k"])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(kmeans_scan["k"], kmeans_scan["Inertia"], "o-", color=PALETTE[0], lw=2, label="Inertia")
axes[0].axvline(elbow_k, color="red", ls="--", label=f"Elbow at k = {elbow_k}")
axes[0].set_title("K-Means Elbow Curve", fontsize=13, weight="bold")
axes[0].set_xlabel("Number of clusters (k)")
axes[0].set_ylabel("Inertia (within-cluster sum of squares)")
axes[0].legend()

axes[1].plot(kmeans_scan["k"], kmeans_scan["Silhouette"], "s-", color=PALETTE[2], lw=2, label="Silhouette Score")
axes[1].axvline(silhouette_best_k, color="grey", ls="--", label=f"Best silhouette at k = {silhouette_best_k}")
axes[1].set_title("Silhouette Score vs. k (cross-check)", fontsize=13, weight="bold")
axes[1].set_xlabel("Number of clusters (k)")
axes[1].set_ylabel("Silhouette Score")
axes[1].legend()

plt.tight_layout()
plt.show()

display(kmeans_scan.round(4))

# ---- Choice of k (made WITHOUT looking at maintenance_risk) ----
K_CLUSTERS = elbow_k
# K_CLUSTERS = 3     # <- uncomment to override the automatic choice after discussing the plots as a team
print(f"Geometric elbow: k = {elbow_k} | highest silhouette: k = {silhouette_best_k}")
print(f"Selected k for both algorithms: {K_CLUSTERS}")
````

### F2.1 — Fit K-Means with the Selected k

````python
kmeans_model = KMeans(n_clusters=K_CLUSTERS, n_init=10, random_state=RANDOM_STATE)
kmeans_labels = kmeans_model.fit_predict(X_cluster)

print("K-Means cluster sizes:")
print(pd.Series(kmeans_labels).value_counts().sort_index().rename("Samples").to_frame().assign(
    Percentage=lambda d: (d["Samples"] / len(kmeans_labels) * 100).round(2)))
print(f"\nFinal inertia: {kmeans_model.inertia_:,.2f} | iterations: {kmeans_model.n_iter_}")
````

## F3 — Agglomerative Hierarchical Clustering, Linkage Comparison & Dendrograms

Agglomerative clustering starts with every point as its own cluster and repeatedly merges the two closest clusters. *How* "closest" is defined is the **linkage strategy**:

| Linkage | Distance between clusters | Typical behaviour |
|---|---|---|
| Ward | increase in within-cluster variance | compact, similar-sized clusters |
| Complete | maximum pairwise distance | compact clusters, sensitive to outliers |
| Average | mean pairwise distance | compromise between single and complete |
| Single | minimum pairwise distance | "chaining" – often one giant cluster plus tiny ones |

All four are fitted with the same *k*, scored with the three metrics plus the **cophenetic correlation** (how faithfully the dendrogram preserves the original pairwise distances). The final Agglomerative model uses the linkage with the best Silhouette Score among those whose smallest cluster holds at least 2 % of the sample, which guards against degenerate solutions where a few outliers form their own clusters.

````python
linkage_methods = ["ward", "complete", "average", "single"]
MIN_CLUSTER_FRACTION = 0.02

pairwise_distances = pdist(X_cluster)
Z_by_method = {}
agglo_labels_by_method = {}
linkage_rows = []

for method in linkage_methods:
    Z_by_method[method] = linkage(X_cluster, method=method)
    coph_corr = cophenet(Z_by_method[method], pairwise_distances)[0]

    labels = AgglomerativeClustering(n_clusters=K_CLUSTERS, linkage=method).fit_predict(X_cluster)
    agglo_labels_by_method[method] = labels

    row = {"Linkage": method, "Cophenetic Corr.": coph_corr,
           "Smallest Cluster (%)": pd.Series(labels).value_counts(normalize=True).min() * 100}
    row.update(evaluate_clustering(X_cluster, labels))
    linkage_rows.append(row)

linkage_comparison = pd.DataFrame(linkage_rows)
display(linkage_comparison.round(4))
linkage_comparison.round(4).to_csv(f"{RESULTS_DIR}/agglomerative_linkage_comparison.csv", index=False)

# ---- Select the linkage (guard against degenerate, outlier-only clusters) ----
valid = linkage_comparison[linkage_comparison["Smallest Cluster (%)"] >= MIN_CLUSTER_FRACTION * 100]
if valid.empty:
    best_linkage = "ward"
    print("No linkage met the minimum-cluster-size rule; defaulting to Ward.")
else:
    best_linkage = valid.loc[valid["Silhouette Score"].idxmax(), "Linkage"]
agglo_labels = agglo_labels_by_method[best_linkage]

print(f"Selected linkage: {best_linkage}")
print("Agglomerative cluster sizes:", pd.Series(agglo_labels).value_counts().sort_index().to_dict())
````

### F3.1 — Dendrograms for All Four Linkage Strategies

**Why we plot this:** a dendrogram records the full merge history. The *height* of a merge is the distance at which two clusters were joined, so a long vertical gap with no merges suggests a natural place to cut the tree. The dashed line marks the cut that yields the selected *k*. Only the last 30 merges are drawn (`truncate_mode="lastp"`); the number in brackets at the bottom is how many samples each leaf represents.

````python
def cut_height(Z, k):
    """Height that cuts the dendrogram into exactly k clusters."""
    return (Z[-(k - 1), 2] + Z[-k, 2]) / 2


# Single linkage "chains" points into a very deep tree and SciPy's dendrogram code is recursive
sys.setrecursionlimit(max(sys.getrecursionlimit(), 20000))
set_link_color_palette(sns.color_palette("colorblind").as_hex())

fig, axes = plt.subplots(2, 2, figsize=(16, 10))
for ax, method in zip(axes.ravel(), linkage_methods):
    Z = Z_by_method[method]
    threshold = cut_height(Z, K_CLUSTERS)
    dendrogram(Z, truncate_mode="lastp", p=30, color_threshold=threshold,
               leaf_rotation=90, leaf_font_size=8, show_contracted=True, ax=ax)
    ax.axhline(threshold, color="black", ls="--", lw=1.3, label=f"Cut for k = {K_CLUSTERS}")
    coph = linkage_comparison.set_index("Linkage").loc[method, "Cophenetic Corr."]
    ax.set_title(f"{method.capitalize()} linkage (cophenetic r = {coph:.3f})", fontsize=12, weight="bold")
    ax.set_xlabel("Cluster (samples in brackets)")
    ax.set_ylabel("Merge distance")
    ax.legend(loc="upper right")
plt.suptitle("Hierarchical Clustering Dendrograms - Linkage Strategy Comparison", fontsize=15, weight="bold", y=1.01)
plt.tight_layout()
plt.show()
````

### F3.2 — Dendrogram of the Selected Linkage

````python
Z_best = Z_by_method[best_linkage]
threshold_best = cut_height(Z_best, K_CLUSTERS)

fig, ax = plt.subplots(figsize=(14, 6))
dendrogram(Z_best, truncate_mode="lastp", p=40, color_threshold=threshold_best,
           leaf_rotation=90, leaf_font_size=8, show_contracted=True, ax=ax)
ax.axhline(threshold_best, color="black", ls="--", lw=1.5, label=f"Cut for k = {K_CLUSTERS}")
ax.set_title(f"Dendrogram - Agglomerative Clustering ({best_linkage.capitalize()} linkage)", fontsize=14, weight="bold")
ax.set_xlabel("Cluster (samples in brackets)")
ax.set_ylabel("Merge distance")
ax.legend()
plt.tight_layout()
plt.show()

set_link_color_palette(None)   # restore SciPy's default palette
````

## F5 — Cluster Visualisation: PCA (2 Components) & t-SNE

### F5.1 — PCA 2-D Scatter Plot for Every Algorithm

**Why we plot this:** the clustering lives in ~20 dimensions, which cannot be drawn directly. PCA projects the data onto the two directions of maximum variance so cluster membership can be inspected visually. Because PCA is linear and keeps only two components, clusters that are separated along other directions can look overlapped. The percentage in each axis label is the variance explained by that component.

````python
pca = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca.fit_transform(X_cluster)
pc1, pc2 = pca.explained_variance_ratio_ * 100
pca_centers = pca.transform(kmeans_model.cluster_centers_)

fig, axes = plt.subplots(1, 2, figsize=(16, 6.5))
scatter_clusters(axes[0], X_pca, kmeans_labels,
                 f"K-Means (k={K_CLUSTERS}) - PCA projection",
                 f"Principal Component 1 ({pc1:.1f}% variance)",
                 f"Principal Component 2 ({pc2:.1f}% variance)", centers=pca_centers)
scatter_clusters(axes[1], X_pca, agglo_labels,
                 f"Agglomerative ({best_linkage}, k={K_CLUSTERS}) - PCA projection",
                 f"Principal Component 1 ({pc1:.1f}% variance)",
                 f"Principal Component 2 ({pc2:.1f}% variance)")
plt.suptitle("Cluster Visualisation with PCA (2 Components)", fontsize=15, weight="bold")
plt.tight_layout()
plt.show()

print(f"Variance captured by the two components: {pc1 + pc2:.1f}% "
      f"(the rest of the structure is not visible in these plots).")
````

### F5.2 — t-SNE Visualisation (Both Algorithms)

**Why we plot this:** t-SNE is a non-linear embedding that preserves *local neighbourhoods*, so it can reveal groups that a linear PCA projection hides. The same embedding is coloured twice, once per algorithm. t-SNE axes have no physical meaning and distances *between* far-apart blobs should not be over-interpreted; it is used here only as a complementary visual check.

````python
tsne = TSNE(n_components=2, perplexity=30, init="pca", learning_rate="auto", random_state=RANDOM_STATE)
X_tsne = tsne.fit_transform(X_cluster)

fig, axes = plt.subplots(1, 2, figsize=(16, 6.5))
scatter_clusters(axes[0], X_tsne, kmeans_labels,
                 f"K-Means (k={K_CLUSTERS}) - t-SNE", "t-SNE dimension 1", "t-SNE dimension 2")
scatter_clusters(axes[1], X_tsne, agglo_labels,
                 f"Agglomerative ({best_linkage}, k={K_CLUSTERS}) - t-SNE", "t-SNE dimension 1", "t-SNE dimension 2")
plt.suptitle("Cluster Visualisation with t-SNE", fontsize=15, weight="bold")
plt.tight_layout()
plt.show()
````

## F6 — Cluster Interpretation & Post-hoc Validation

### F6.1 — Cluster Profiles (Standardised Feature Means)

**Why we plot this:** to describe what makes each cluster different. Each cell shows the cluster's mean of a feature in standard-deviation units (z-score): positive (red) means above the overall average, negative (blue) below. Features with the strongest colours characterise that cluster.

````python
scaled_df = pd.DataFrame(X_cluster, columns=cluster_features)  # winsorised + standardised
profile_kmeans = scaled_df.groupby(kmeans_labels).mean().T
profile_agglo = scaled_df.groupby(agglo_labels).mean().T
profile_kmeans.columns = [f"Cluster {c}" for c in profile_kmeans.columns]
profile_agglo.columns = [f"Cluster {c}" for c in profile_agglo.columns]

fig, axes = plt.subplots(1, 2, figsize=(16, 9))
sns.heatmap(profile_kmeans, cmap="coolwarm", center=0, annot=True, fmt=".2f", ax=axes[0], cbar_kws={"label": "Mean z-score"})
axes[0].set_title(f"K-Means (k={K_CLUSTERS}) - Cluster Profiles", fontsize=13, weight="bold")
axes[0].set_xlabel("Cluster")
axes[0].set_ylabel("Feature")
sns.heatmap(profile_agglo, cmap="coolwarm", center=0, annot=True, fmt=".2f", ax=axes[1], cbar_kws={"label": "Mean z-score"})
axes[1].set_title(f"Agglomerative ({best_linkage}, k={K_CLUSTERS}) - Cluster Profiles", fontsize=13, weight="bold")
axes[1].set_xlabel("Cluster")
axes[1].set_ylabel("Feature")
plt.tight_layout()
plt.show()

# Most distinguishing features per K-Means cluster (largest absolute z-score)
for c in profile_kmeans.columns:
    top3 = profile_kmeans[c].abs().sort_values(ascending=False).head(3).index
    desc = ", ".join(f"{f} ({profile_kmeans.loc[f, c]:+.2f})" for f in top3)
    print(f"{c}: {desc}")
````

### F6.2 — Post-hoc Validation Against `maintenance_risk`

Only **now** are the ground-truth risk labels brought in. Two external metrics measure agreement between clusters and labels, regardless of how clusters are numbered:

* **Adjusted Rand Index (ARI):** 1 = identical partitions, ≈ 0 = no better than random labelling.
* **Normalised Mutual Information (NMI):** 0 = clusters carry no information about the labels, 1 = perfect agreement.

The cross-tabulation shows, for each cluster, the share of Low / Medium / High risk machines.

````python
validation_rows = []
for name, labels in [(f"K-Means (k={K_CLUSTERS})", kmeans_labels),
                     (f"Agglomerative ({best_linkage}, k={K_CLUSTERS})", agglo_labels)]:
    validation_rows.append({
        "Algorithm": name,
        "Adjusted Rand Index": adjusted_rand_score(risk_labels_for_validation, labels),
        "Normalised Mutual Info": normalized_mutual_info_score(risk_labels_for_validation, labels),
    })
validation_df = pd.DataFrame(validation_rows)
display(validation_df.round(4))
validation_df.round(4).to_csv(f"{RESULTS_DIR}/clustering_external_validation.csv", index=False)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
for ax, (name, labels) in zip(axes, [(f"K-Means (k={K_CLUSTERS})", kmeans_labels),
                                     (f"Agglomerative ({best_linkage})", agglo_labels)]):
    ct = pd.crosstab(labels, risk_labels_for_validation.values, normalize="index") * 100
    ct = ct[[c for c in ["Low", "Medium", "High"] if c in ct.columns]]
    ct.index = [f"Cluster {c}" for c in ct.index]
    sns.heatmap(ct, annot=True, fmt=".1f", cmap="Blues", vmin=0, vmax=100, ax=ax, cbar_kws={"label": "% of cluster"})
    ax.set_title(f"{name}: risk mix per cluster", fontsize=12, weight="bold")
    ax.set_xlabel("maintenance_risk (ground truth, used only after fitting)")
    ax.set_ylabel("Cluster")
plt.tight_layout()
plt.show()

overall = risk_labels_for_validation.value_counts(normalize=True).mul(100).round(1).to_dict()
print("Overall risk mix in the sample:", overall)
print("Interpretation guide: if a cluster's risk mix differs clearly from the overall mix, the cluster is "
      "informative about maintenance risk; if it matches the overall mix, it is not.")
````

## H1 — Export the Regression Model and App Metadata

Before exporting, the regression pipeline is re-scored on the test split and compared with the Section C result, to make sure the saved model is exactly the one evaluated earlier. The metadata file lets the app build its input form (valid ranges, drop-down options, typical default values) without needing the dataset.

````python
import json
import sys
import sklearn
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# ---- 1) Sanity check: the exported regression pipeline must reproduce the Section C result ----
reg_test_pred = best_gb_tuned.predict(reg_preprocessor.transform(X_reg_test))
reg_r2 = float(r2_score(y_reg_test, reg_test_pred))
reg_rmse = float(np.sqrt(mean_squared_error(y_reg_test, reg_test_pred)))
reg_mae = float(mean_absolute_error(y_reg_test, reg_test_pred))
print(f"Regression pipeline R2 on the test split: {reg_r2:.4f} (Section C reported {gb_tuned_r2:.4f})")
if abs(reg_r2 - gb_tuned_r2) > 1e-6:
    print("WARNING: the values differ - re-run Sections C and the preprocessing cells in order before exporting.")

# ---- 2) Save the regression bundle ----
joblib.dump(
    {
        "model": best_gb_tuned,
        "preprocessor": reg_preprocessor,
        "feature_columns": X_reg.columns.tolist(),
        "model_name": "Gradient Boosting Regressor (tuned)",
    },
    f"{MODELS_DIR}/final_regressor.pkl",
    compress=3,
)

# ---- 3) Metadata for the app's input form ----
feature_meta = {"numeric": {}, "categorical": {}}
for col in numerical_features:
    s = X_class[col]
    feature_meta["numeric"][col] = {
        "min": float(s.min()), "max": float(s.max()), "median": float(s.median()),
        "p01": float(s.quantile(0.01)), "p99": float(s.quantile(0.99)),
        "is_integer": bool(np.all(np.isclose(s % 1, 0))),
    }
for col in categorical_features:
    s = X_class[col].astype(str)
    feature_meta["categorical"][col] = {"options": sorted(s.unique().tolist()), "default": str(s.mode().iloc[0])}

clf_row = comparison_df.set_index("Algorithm").loc[best_name]
app_metadata = {
    "class_labels": label_encoder.classes_.tolist(),
    "classifier": {
        "name": best_name,
        "accuracy": float(clf_row["Accuracy"]),
        "weighted_f1": float(clf_row["F1-score (Weighted)"]),
        "roc_auc": float(clf_row["ROC-AUC (OvR)"]),
    },
    "regressor": {"name": "Gradient Boosting Regressor (tuned)", "r2": reg_r2, "rmse": reg_rmse, "mae": reg_mae},
    "features": feature_meta,
    "versions": {
        "python": sys.version.split()[0],
        "scikit-learn": sklearn.__version__,
        "numpy": np.__version__,
        "pandas": pd.__version__,
    },
}
with open(f"{MODELS_DIR}/app_metadata.json", "w", encoding="utf-8") as f:
    json.dump(app_metadata, f, indent=2)

# ---- 4) Report what was written (GitHub rejects single files above 100 MB) ----
for fname in ["final_classifier.pkl", "final_regressor.pkl", "app_metadata.json"]:
    size_mb = os.path.getsize(os.path.join(MODELS_DIR, fname)) / 1e6
    flag = "   <-- too large for GitHub (100 MB limit)" if size_mb > 95 else ""
    print(f"{MODELS_DIR}/{fname}: {size_mb:.2f} MB{flag}")
print("\nNumeric inputs for the app    :", len(feature_meta["numeric"]))
print("Categorical inputs for the app:", len(feature_meta["categorical"]))
````

## H2 — Preview: What the App Will Compute

This cell repeats, in a few lines, exactly what `app/app.py` does when a visitor presses **Predict**: build one input row (typical values from the data), apply the saved preprocessors, and call the saved models. It confirms the exported files load and work together before anything is deployed.

````python
# Load the exported files exactly as the app does
clf_bundle = joblib.load(f"{MODELS_DIR}/final_classifier.pkl")
reg_bundle = joblib.load(f"{MODELS_DIR}/final_regressor.pkl")
with open(f"{MODELS_DIR}/app_metadata.json", encoding="utf-8") as f:
    meta_check = json.load(f)

# One example input: the typical (median / most frequent) value of every field
example_input = {c: m["median"] for c, m in meta_check["features"]["numeric"].items()}
example_input.update({c: m["default"] for c, m in meta_check["features"]["categorical"].items()})
example_row = pd.DataFrame([example_input])
example_row["bearing_temp_rise_c"] = example_row["bearing_temperature_c"] - example_row["ambient_temperature_c"]

risk_proba = clf_bundle["model"].predict_proba(clf_bundle["preprocessor"].transform(example_row[clf_bundle["feature_columns"]]))[0]
energy_kwh = reg_bundle["model"].predict(reg_bundle["preprocessor"].transform(example_row[reg_bundle["feature_columns"]]))[0]

print("Example input (typical machine):")
display(example_row.T.rename(columns={0: "Value"}))
print(f"\nPredicted maintenance risk : {clf_bundle['label_encoder'].classes_[risk_proba.argmax()]}")
print("Class probabilities        :", {c: round(float(p), 3) for c, p in zip(clf_bundle["label_encoder"].classes_, risk_proba)})
print(f"Predicted energy use       : {energy_kwh:.2f} kWh")
````
